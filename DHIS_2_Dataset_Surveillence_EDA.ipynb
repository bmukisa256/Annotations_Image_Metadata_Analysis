```python
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import linregress

# Create the regression plot
sns.regplot(x='Presumptive TB patients with documented HIV positive results(Underlying conditions) ',
            y='Died Male', data=DHIS2_TB_2021, scatter_kws={'alpha': 0.5})

# Calculate the slope (m) and intercept (c) of the regression line
slope, intercept, r_value, p_value, std_err = linregress(
    DHIS2_TB_2021['Presumptive TB patients with documented HIV positive results(Underlying conditions) '],
    DHIS2_TB_2021['Died Male']
)

# Display the equation of the line on the plot
equation = f'y = {slope:.2f}x + {intercept:.2f}'
plt.text(0.1, 0.9, equation, transform=plt.gca().transAxes, fontsize=12, color='red')

# Add labels and title
plt.xlabel('Presumptive TB patients with documented HIV positive results (Underlying conditions) ')
plt.ylabel('Died Male')
plt.title('Regression plot of Presumptive TB patients with documented HIV positive results (Underlying conditions) vs. Died Male for 2021')

# Show the plot
plt.show()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

Year = ("2020", "2021", "2022", "2023", "2024")
TB_data = {
    'Total Died': (25200, 27474, 4369, 4338, 3677),
    'Total Cured': (20566, 23143, 29288, 43847, 46654),
    'HIV/TB': (96191, 120349, 150151, 141240, 155856),
}

x = np.arange(len(Year))  # the label locations
width = 0.25  # the width of the bars
multiplier = 0

fig, ax = plt.subplots(layout='constrained')

for attribute, measurement in TB_data.items():
    offset = width * multiplier
    rects = ax.bar(x + offset, measurement, width, label=attribute)
    ax.bar_label(rects, padding=3)
    multiplier += 1

# Add some text for labels, title and custom x-axis tick labels
ax.set_ylabel('cases')
ax.set_title(' TB data from 2020 to 2024')
ax.set_xticks(x + width, Year)
ax.legend(loc='upper left', ncols=3)
ax.set_ylim(0, 160000)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Ensure the 'Region' column is consistent across all dataframes
regions = DHIS2_TB_2020['Region '].unique()  # Get unique regions

# Define the years
years = [2020, 2021, 2022, 2023, 2024]

# Define the data for each year
data = {
    2020: DHIS2_TB_2020.groupby('Region ')['Died Male'].sum(),
    2021: DHIS2_TB_2021.groupby('Region ')['Died Male'].sum(),
    2022: DHIS2_TB_2022.groupby('Region ')['Died Male '].sum(),
    2023: DHIS2_TB_2023.groupby('Region ')['Died Male'].sum(),
    2024: DHIS2_TB_2024.groupby('Region ')['Died Male'].sum()
}

# Define colors for each year
colors = ['red', 'blue', 'green', 'orange', 'purple']

# Initialize the bottom array to zero
bottom = np.zeros(len(regions))

# Plot the bars
plt.figure(figsize=(17, 6))

# Loop through each year and plot the bars
for i, year in enumerate(years):
    values = data[year].reindex(regions, fill_value=0).values  # Get values for the current year, aligning with regions
    plt.bar(regions, values, bottom=bottom, alpha=0.5, color=colors[i], label=str(year))
    bottom += values  # Update the bottom for the next year

# Add labels and legend
plt.xlabel('Region')
plt.ylabel('Number of Deaths')
plt.title('TB Deaths by Region and Year (Male)')
plt.xticks(rotation=45)  # Rotate x-axis labels for better readability
plt.legend()

# Adjust layout to prevent clipping of labels
plt.tight_layout()

plt.show()
```
